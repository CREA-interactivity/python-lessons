## 2. Loops with a counter 🔢
`range(5)` counts **0, 1, 2, 3, 4**. Each time around, the box `i` holds the next number — and we can use it!

```python
for i in range(5):
    print("Count:", i)
```

Because `i` is a real number, we can do math with it:

```python
for i in range(5):
    print(i, "doubled is", i * 2)
```

## 3. Bigger choices — `if` / `elif` / `else` 🔀
`if` checks the first thing. `elif` means *"or else, check THIS"*. `else` is what happens when nothing above was true. Python stops at the **first** one that's true.

```python
weather = "rainy"

if weather == "sunny":
    print("Wear sunglasses 😎")
elif weather == "rainy":
    print("Grab an umbrella ☔")
elif weather == "snowy":
    print("Build a snowman ⛄")
else:
    print("Hmm, I'm not sure what that is!")
```

## 4. 🐢 Meet the turtle!
The turtle is a little robot that leaves a line wherever it walks. You steer it:
- `forward(100)` — walk forward 100 steps
- `left(90)` / `right(90)` — turn (90 is a square corner)

```python
import turtle

t = turtle.Turtle()
t.speed(3)          # 1 = slow, 10 = fast, 0 = zoom!

t.forward(100)
t.left(90)
t.forward(100)

turtle.done()
```

## 5. Colors & pen size 🎨
The turtle can change color and line thickness. `right` turns the other way, and you can lift the pen to move **without** drawing.

```python
import turtle

t = turtle.Turtle()
t.speed(3)

t.pensize(5)              # thicker line
t.pencolor("purple")      # try "red", "blue", "green", "orange"...
t.forward(120)

t.right(90)
t.pencolor("orange")
t.forward(120)

turtle.done()
```

**🎯 Your turn:** Change the two colors, and make the line even thicker with `pensize`.

## 6. Loops + turtle = superpowers 🚀
Drawing a square meant repeating the same two lines 4 times. A **loop** does the repeating for us!

```python
import turtle

t = turtle.Turtle()
t.speed(5)

for side in range(4):     # do this 4 times
    t.forward(100)
    t.right(90)           # 90 makes square corners

turtle.done()
```

**🎯 Your turn:** Change `range(4)` to `range(3)` **and** `right(90)` to `right(120)`. What shape appears? 🔺

Now the magic trick: draw that square **many times**, turning a little each time. Watch a flower appear! 🌸

```python
import turtle

t = turtle.Turtle()
t.speed(0)                # fastest, so we don't wait forever
t.pencolor("deeppink")

for flower in range(36):      # spin around 36 times
    for side in range(4):     # draw one square
        t.forward(100)
        t.right(90)
    t.right(10)               # turn a little, then draw the next square

turtle.done()
```
